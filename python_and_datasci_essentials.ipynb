### Lambda, Map, Filter, Reduce

````python
from functools import reduce

nums = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, nums)) # applies a lambda expression to each element
evens = list(filter(lambda x: x % 2 == 0, nums)) # filters to a subset based on conditionals

# Reduce is a "rolling computation"
add_lambda = lambda x,y: x+y
total = reduce(add_lambda, nums) # first pairing x=1,y=2, then x=prev_result & y=next_value; it's 1,2; 3,3; 6,4; return is int 10

print(f'''
add: {add} | type: {type(add)}
squared: {squared} | type: {type(squared)}
evens: {evens} | type: {type(evens)}
total: {total} | type: {type(total)}
''')
````

### Type Hinting & Docstring

````python
# Notice the param type hints, the param default value, the return type, and the docstring (displayed with shift-tab while cursor inside the parentheses of invoking fxn call
def greet_users(names: list[str], excited: bool = False) -> list[str]:
    """
    Returns a list of greeting messages for each name.

    Args:
        names (list[str]): A list of user names to greet.
        excited (bool, optional): If True, adds exclamation to the greeting. Defaults to False.

    Returns:
        list[str]: A list of personalized greeting messages.
    """
    return [f"Hello, {name}{'!' if excited else ''}" for name in names]
````

````python
# Example usage
greet_users(names=['Jake','Adam','Eve'], excited=True)
````

````python
from typing import List, Dict, Tuple, Union, Optional

def process_user_data(
    names: List[str],
    ages: Dict[str, int],
    active: bool = True,
    location: Optional[str] = None,
    score_range: Union[int, Tuple[int, int]] = (0, 100),
    preferences: Optional[Dict[str, str]] = None
) -> Dict[str, Union[int, str]]:
    """
    Process user data and return a summary dictionary.

    Args:
        names (List[str]): List of user names.
        ages (Dict[str, int]): Dictionary of user names to their ages.
        active (bool, optional): Whether to include only active users. Defaults to True.
        location (Optional[str], optional): Location to tag to all users. Defaults to None.
        score_range (Union[int, Tuple[int, int]], optional): Score or score range filter. Defaults to (0, 100).
        preferences (Optional[Dict[str, str]], optional): User preferences dictionary. Defaults to None.

    Returns:
        Dict[str, Union[int, str]]: A summary including user count and status.
    """
    summary = {
        "user_count": len(names),
        "location": location if location is not None else "unspecified",
        "active_status": "active" if active else "inactive"
    }

    if preferences is not None:
        summary.update(preferences)

    if isinstance(score_range, tuple):
        summary["score_range"] = f"{score_range[0]}–{score_range[1]}"
    else:
        summary["score_range"] = str(score_range)

    return summary
````

````python
# Example usage
example_summary = process_user_data(
    names=["Alice", "Bob"],
    ages={"Alice": 30, "Bob": 25},
    active=False,
    location="NYC",
    score_range=75,
    preferences={"theme": "dark"}
)

print(example_summary)
````

### Generators (yield)

````python
def my_generator(n):
    for i in range(n):
        yield i # the yield makes the fxn a generator to pause execution & make an iterable object

gen = my_generator(5)
print(next(gen)) # Output: 0
print(next(gen)) # Output: 1

for item in gen:
    print(item) # Output: 2 3 4
````

### Decorators & Context Managers (with)

````python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before")
        result = func(*args, **kwargs)
        print("After")
        return result
    return wrapper

@my_decorator
def greet(name):
    print(f"Hello {name}")

greet("Jake")
````

````python
from contextlib import contextmanager

@contextmanager # allows defining a context manager using a generator fxn (i.e. a fxn w/ yield)
def my_context():
    print("Entering") # setup code; pre-execution
    yield # context is handed over to the block inside the with statement; yield divides code into two parts (pre & post)
    print("Exiting") # cleanup code; post-execution

with my_context(): # even a with open(file): is a context manager
    print("Inside block")
````

### Custom Exceptions / Error Handling

````python
class NegativeValueError(ValueError):
    """Custom exception raised when a negative value is encountered where not allowed."""
    pass

def square_root(value: float) -> float:
    if value < 0:
        raise NegativeValueError("Cannot take the square root of a negative number.")
    return value ** 0.5

# Example usage:
try:
    print(square_root(16))   # 4.0
    print(square_root(-9))   # Raises NegativeValueError
except NegativeValueError as e:
    print(f"Error: {e}")
````

### Unpacking / Star Expressions

````python
# Standard tuple unpacking
name, age = ("Jake", 30)

# Extended unpacking with *
first, *middle, last = [1, 2, 3, 4, 5]

# Swapping values without a temp variable
a, b = 5, 10
a, b = b, a  # a = 10, b = 5

print("Name:", name)
print("Age:", age)
print("First:", first)
print("Middle:", middle)
print("Last:", last)
print("Swapped: a =", a, ", b =", b)
````

### Regex

<details><summary><strong>Common Regex Symbols Reference</strong></summary>

| Pattern        | Description                                          |
|----------------|------------------------------------------------------|
| `.`            | Any character except newline                         |
| `^`            | Start of string (or start of line with `re.MULTILINE`) |
| `$`            | End of string (or end of line with `re.MULTILINE`)   |
| `*`            | 0 or more of the preceding element                   |
| `+`            | 1 or more of the preceding element                   |
| `?`            | 0 or 1 of the preceding element (also makes quantifiers non-greedy) |
| `{n}`          | Exactly n repetitions                                |
| `{n,}`         | n or more repetitions                                |
| `{n,m}`        | Between n and m repetitions                          |
| `[]`           | Set of characters (e.g., `[aeiou]`)                  |
| `[^]`          | Negated set (e.g., `[^0-9]` means not a digit)       |
| `\d`           | Digit character (`[0-9]`)                            |
| `\D`           | Non-digit character                                  |
| `\w`           | Word character (`[a-zA-Z0-9_]`)                      |
| `\W`           | Non-word character                                   |
| `\s`           | Whitespace character                                 |
| `\S`           | Non-whitespace character                             |
| `|`            | Alternation (OR) — e.g., `cat|dog`                   |
| `(...)`        | Capturing group                                      |
| `(?:...)`      | Non-capturing group                                  |
| `(?P<name>...)`| Named capturing group                                |
| `(?=...)`      | Positive lookahead                                   |
| `(?!...)`      | Negative lookahead                                   |

Use `re.MULTILINE`, `re.IGNORECASE`, and other flags as needed to adjust matching behavior.
</details>

````python
import re

# --- re.search --- #
text = "My phone number is 123-456-7890."
pattern = r"\d{3}-\d{3}-\d{4}" # \d matches any digit, {3} means "3 times"
match = re.search(pattern, text)
if match:
    print("Phone number found:", match.group())

# --- re.search w/ Groups --- #
text = "Name: John, Age: 30"
pattern = r"Name: (?P<name>\w+), Age: (?P<age>\d+)" # the names & saved matches become key-value pairs where value is returned via match.group(key)
match = re.search(pattern, text)
if match:
    print("Name:", match.group("name"))
    print("Age:", match.group("age"))

# --- re.findall --- #
text = "Emails: test@example.com, hello@world.net"
pattern = r"[\w.-_]+@[\w]+\.\w+" # any number of word,dots,hyphens,digits(_) then @ then any # of word chars then dot then any # of word chars
emails = re.findall(pattern, text)
print(emails)
````

````python
import numpy as np

# np.array() - Create a NumPy array from a list or tuple
arr = np.array([1, 2, 3])
print("np.array():", arr)  # Output: [1 2 3]

# np.arange() - Create an array with evenly spaced values
arr = np.arange(0, 10, 2) # focused on step size
print("np.arange():", arr)  # Output: [0 2 4 6 8]

# np.linspace() - Create an array with evenly spaced values over a specified interval
arr = np.linspace(0, 1, 5) # focused on # of elements in the arr
print("np.linspace():", arr)  # Output: [0.   0.25 0.5  0.75 1.  ]
````
